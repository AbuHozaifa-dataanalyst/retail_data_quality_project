```python
product_data = []

subcategory_list = [
    (category, subcategory)
    for category, subcategories in categories.items()
    for subcategory in subcategories
]

for i in range(1, 501):

    category, subcategory = subcategory_list[
        np.random.randint(len(subcategory_list))
    ]

    brand = np.random.choice(brands)

    template = np.random.choice(
        product_templates[subcategory]
    )

    product_name = f"{brand} {template} {i:03d}"

    unit_cost = round(
        np.random.uniform(20, 3000),
        2
    )

    markup = np.random.uniform(1.15, 1.60)

    selling_price = round(
        unit_cost * markup,
        2
    )

    product_data.append({
        "product_id": f"P{i:04d}",
        "product_name": product_name,
        "category": category,
        "subcategory": subcategory,
        "brand": brand,
        "unit_cost": unit_cost,
        "selling_price": selling_price
    })

dim_product = pd.DataFrame(product_data)

dim_product.head()
```

```python
N_TRANSACTIONS = 120_000

print(f"Generating {N_TRANSACTIONS:,} transactions")
```

```python
date_range = pd.date_range(
    start="2024-01-01",
    end="2025-12-31",
    freq="D"
)

transaction_dates = np.random.choice(
    date_range,
    size=N_TRANSACTIONS
)
```

```python
transaction_product_ids = np.random.choice(
    dim_product["product_id"],
    size=N_TRANSACTIONS
)
```

```python
transaction_product_ids = np.random.choice(
    dim_product["product_id"],
    size=N_TRANSACTIONS
)
```

```python
transaction_customer_ids = np.random.choice(
    dim_customer["customer_id"],
    size=N_TRANSACTIONS
)
```

```python
transaction_store_ids = np.random.choice(
    dim_store["store_id"],
    size=N_TRANSACTIONS
)
```

```python
payment_methods = [
    "Card",
    "Cash",
    "Mada",
    "Apple Pay",
    "STC Pay"
]
```

```python
transaction_payment_methods = np.random.choice(
    payment_methods,
    size=N_TRANSACTIONS,
    p=[0.35, 0.15, 0.25, 0.15, 0.10]
)
```

```python
sales_channels = [
    "Store",
    "E-commerce"
]
```

```python
transaction_channels = np.random.choice(
    sales_channels,
    size=N_TRANSACTIONS,
    p=[0.70, 0.30]
)
```

```python
quantities = np.random.poisson(
    lam=2,
    size=N_TRANSACTIONS
) + 1

quantities = np.clip(
    quantities,
    1,
    10
)
```

```python
product_lookup = dim_product[
    [
        "product_id",
        "unit_cost",
        "selling_price"
    ]
].copy()
```

```python
transactions = pd.DataFrame({
    "transaction_id": [
        f"T{i:06d}"
        for i in range(1, N_TRANSACTIONS + 1)
    ],
    "transaction_date": transaction_dates,
    "customer_id": transaction_customer_ids,
    "product_id": transaction_product_ids,
    "store_id": transaction_store_ids,
    "quantity": quantities,
    "payment_method": transaction_payment_methods,
    "sales_channel": transaction_channels
})
```

```python
transactions = transactions.merge(
    product_lookup,
    on="product_id",
    how="left"
)
```

```python
price_variation = np.random.uniform(
    0.95,
    1.05,
    size=N_TRANSACTIONS
)

transactions["unit_price"] = (
    transactions["selling_price"] * price_variation
).round(2)
```

```python
transactions["gross_sales"] = (
    transactions["quantity"]
    * transactions["unit_price"]
).round(2)
```

```python
transactions["gross_sales"] = (
    transactions["quantity"]
    * transactions["unit_price"]
).round(2)
```

```python
discount_rate = np.random.choice(
    [0, 0.05, 0.10, 0.15, 0.20],
    size=N_TRANSACTIONS,
    p=[0.45, 0.25, 0.18, 0.08, 0.04]
)
```

```python
transactions["discount_amount"] = (
    transactions["gross_sales"]
    * discount_rate
).round(2)
```

```python
transactions["net_sales"] = (
    transactions["gross_sales"]
    - transactions["discount_amount"]
).round(2)
```

```python
transactions["cost_amount"] = (
    transactions["quantity"]
    * transactions["unit_cost"]
).round(2)
```

```python
transactions["profit_amount"] = (
    transactions["net_sales"]
    - transactions["cost_amount"]
).round(2)
```

```python
transactions["gross_margin_pct"] = np.where(
    transactions["net_sales"] != 0,
    (
        transactions["profit_amount"]
        / transactions["net_sales"]
        * 100
    ),
    np.nan
)

transactions["gross_margin_pct"] = (
    transactions["gross_margin_pct"]
    .round(2)
)
```

```python
transactions = transactions.drop(
    columns=["selling_price"]
)
```

```python
transactions = transactions[
    [
        "transaction_id",
        "transaction_date",
        "customer_id",
        "product_id",
        "store_id",
        "quantity",
        "unit_price",
        "discount_amount",
        "gross_sales",
        "net_sales",
        "unit_cost",
        "cost_amount",
        "profit_amount",
        "gross_margin_pct",
        "payment_method",
        "sales_channel"
    ]
]
```

```python
gross_sales_check = (
    transactions["quantity"]
    * transactions["unit_price"]
).round(2)

print(
    np.isclose(
        transactions["gross_sales"],
        gross_sales_check
    ).all()
)
```

```python
net_sales_check = (
    transactions["gross_sales"]
    - transactions["discount_amount"]
).round(2)

print(
    np.isclose(
        transactions["net_sales"],
        net_sales_check
    ).all()
)
```

```python
profit_check = (
    transactions["net_sales"]
    - transactions["cost_amount"]
).round(2)

print(
    np.isclose(
        transactions["profit_amount"],
        profit_check
    ).all()
)
```

```python
messy_sales = transactions.copy()

print("Original rows:", len(messy_sales))
```

```python
rng = np.random.default_rng(42)
```

```python
corruption_log = []
```

```python
n_missing_customer = int(len(messy_sales) * 0.005)

missing_customer_idx = rng.choice(
    messy_sales.index,
    size=n_missing_customer,
    replace=False
)

messy_sales.loc[
    missing_customer_idx,
    "customer_id"
] = np.nan

corruption_log.append({
    "issue": "Missing customer_id",
    "records_affected": n_missing_customer
})
```

```python
n_missing_payment = int(len(messy_sales) * 0.003)

missing_payment_idx = rng.choice(
    messy_sales.index,
    size=n_missing_payment,
    replace=False
)

messy_sales.loc[
    missing_payment_idx,
    "payment_method"
] = np.nan

corruption_log.append({
    "issue": "Missing payment_method",
    "records_affected": n_missing_payment
})
```

```python
n_duplicates = int(len(messy_sales) * 0.004)

duplicate_rows = messy_sales.sample(
    n=n_duplicates,
    random_state=42
)

messy_sales = pd.concat(
    [messy_sales, duplicate_rows],
    ignore_index=True
)

corruption_log.append({
    "issue": "Exact duplicate transactions",
    "records_affected": n_duplicates
})
```

```python
n_duplicate_ids = 200

duplicate_id_source = messy_sales.sample(
    n=n_duplicate_ids,
    random_state=100
)
```

```python
duplicate_id_rows = duplicate_id_source.copy()

duplicate_id_rows["quantity"] = (
    duplicate_id_rows["quantity"] + 1
)

duplicate_id_rows["gross_sales"] = (
    duplicate_id_rows["quantity"]
    * duplicate_id_rows["unit_price"]
).round(2)

duplicate_id_rows["net_sales"] = (
    duplicate_id_rows["gross_sales"]
    - duplicate_id_rows["discount_amount"]
).round(2)

duplicate_id_rows["cost_amount"] = (
    duplicate_id_rows["quantity"]
    * duplicate_id_rows["unit_cost"]
).round(2)

duplicate_id_rows["profit_amount"] = (
    duplicate_id_rows["net_sales"]
    - duplicate_id_rows["cost_amount"]
).round(2)
```

```python
messy_sales = pd.concat(
    [messy_sales, duplicate_id_rows],
    ignore_index=True
)

corruption_log.append({
    "issue": "Duplicate transaction IDs",
    "records_affected": n_duplicate_ids
})
```

```python
n_negative_qty = 300

negative_qty_idx = rng.choice(
    messy_sales.index,
    size=n_negative_qty,
    replace=False
)

messy_sales.loc[
    negative_qty_idx,
    "quantity"
] = -messy_sales.loc[
    negative_qty_idx,
    "quantity"
].abs()

corruption_log.append({
    "issue": "Negative quantity",
    "records_affected": n_negative_qty
})
```

```python
n_zero_qty = 200

zero_qty_idx = rng.choice(
    messy_sales.index,
    size=n_zero_qty,
    replace=False
)

messy_sales.loc[
    zero_qty_idx,
    "quantity"
] = 0

corruption_log.append({
    "issue": "Zero quantity",
    "records_affected": n_zero_qty
})
```

```python
n_negative_price = 150

negative_price_idx = rng.choice(
    messy_sales.index,
    size=n_negative_price,
    replace=False
)

messy_sales.loc[
    negative_price_idx,
    "unit_price"
] = -messy_sales.loc[
    negative_price_idx,
    "unit_price"
].abs()

corruption_log.append({
    "issue": "Negative unit price",
    "records_affected": n_negative_price
})
```

```python
n_invalid_discount = 200

invalid_discount_idx = rng.choice(
    messy_sales.index,
    size=n_invalid_discount,
    replace=False
)

messy_sales.loc[
    invalid_discount_idx,
    "discount_amount"
] = (
    messy_sales.loc[
        invalid_discount_idx,
        "gross_sales"
    ].abs()
    * rng.uniform(1.10, 1.50, size=n_invalid_discount)
).round(2)

corruption_log.append({
    "issue": "Discount greater than gross sales",
    "records_affected": n_invalid_discount
})
```

```python
n_wrong_gross_sales = 250

wrong_gross_idx = rng.choice(
    messy_sales.index,
    size=n_wrong_gross_sales,
    replace=False
)

messy_sales.loc[
    wrong_gross_idx,
    "gross_sales"
] = (
    messy_sales.loc[
        wrong_gross_idx,
        "gross_sales"
    ] * 1.10
).round(2)

corruption_log.append({
    "issue": "Incorrect gross_sales calculation",
    "records_affected": n_wrong_gross_sales
})
```

```python
n_wrong_profit = 250

wrong_profit_idx = rng.choice(
    messy_sales.index,
    size=n_wrong_profit,
    replace=False
)

messy_sales.loc[
    wrong_profit_idx,
    "profit_amount"
] = (
    messy_sales.loc[
        wrong_profit_idx,
        "profit_amount"
    ] * 0.80
).round(2)

corruption_log.append({
    "issue": "Incorrect profit calculation",
    "records_affected": n_wrong_profit
})
```

```python
n_future_dates = 100

future_idx = rng.choice(
    messy_sales.index,
    size=n_future_dates,
    replace=False
)

future_dates = pd.date_range(
    start="2026-01-01",
    end="2026-12-31",
    periods=n_future_dates
)

messy_sales.loc[
    future_idx,
    "transaction_date"
] = future_dates

corruption_log.append({
    "issue": "Future transaction date",
    "records_affected": n_future_dates
})
```

```python
n_missing_dates = 150

missing_date_idx = rng.choice(
    messy_sales.index,
    size=n_missing_dates,
    replace=False
)

messy_sales.loc[
    missing_date_idx,
    "transaction_date"
] = np.nan

corruption_log.append({
    "issue": "Missing transaction date",
    "records_affected": n_missing_dates
})
```

```python
n_orphan_products = 150

orphan_product_idx = rng.choice(
    messy_sales.index,
    size=n_orphan_products,
    replace=False
)

messy_sales.loc[
    orphan_product_idx,
    "product_id"
] = [
    f"P9{rng.integers(100, 999)}"
    for _ in range(n_orphan_products)
]

corruption_log.append({
    "issue": "Orphan product ID",
    "records_affected": n_orphan_products
})
```

```python
n_orphan_customers = 150

orphan_customer_idx = rng.choice(
    messy_sales.index,
    size=n_orphan_customers,
    replace=False
)

messy_sales.loc[
    orphan_customer_idx,
    "customer_id"
] = [
    f"C9{rng.integers(10000, 99999)}"
    for _ in range(n_orphan_customers)
]

corruption_log.append({
    "issue": "Orphan customer ID",
    "records_affected": n_orphan_customers
})
```

```python
n_orphan_stores = 100

orphan_store_idx = rng.choice(
    messy_sales.index,
    size=n_orphan_stores,
    replace=False
)

messy_sales.loc[
    orphan_store_idx,
    "store_id"
] = [
    f"S{rng.integers(900, 999)}"
    for _ in range(n_orphan_stores)
]

corruption_log.append({
    "issue": "Orphan store ID",
    "records_affected": n_orphan_stores
})
```

```python
n_text_prices = 100

text_price_idx = rng.choice(
    messy_sales.index,
    size=n_text_prices,
    replace=False
)

messy_sales.loc[
    text_price_idx,
    "unit_price"
] = (
    "SAR "
    + messy_sales.loc[
        text_price_idx,
        "unit_price"
    ].abs().round(2).astype(str)
)

corruption_log.append({
    "issue": "Unit price stored as text/currency string",
    "records_affected": n_text_prices
})
```

```python
category_lookup = dim_product[
    ["product_id", "category"]
].drop_duplicates()

messy_sales = messy_sales.merge(
    category_lookup,
    on="product_id",
    how="left"
)
```

```python
category_variants = {
    "Electronics": ["Electronics", "electronics", "ELECTRONICS", "Electronic"],
    "Home Appliances": ["Home Appliances", "home appliances", "HOME APPLIANCES"],
    "Fashion": ["Fashion", "fashion", "FASHION"],
    "Beauty": ["Beauty", "beauty", "BEAUTY"],
    "Sports": ["Sports", "sports", "SPORT"],
    "Grocery": ["Grocery", "grocery", "GROCERY"],
    "Home & Living": ["Home & Living", "home & living", "HOME & LIVING"]
}
```

```python
n_inconsistent_categories = 500

category_idx = rng.choice(
    messy_sales.index,
    size=n_inconsistent_categories,
    replace=False
)

for idx in category_idx:
    current_category = messy_sales.loc[idx, "category"]

    if current_category in category_variants:
        messy_sales.loc[idx, "category"] = rng.choice(
            category_variants[current_category]
        )

corruption_log.append({
    "issue": "Inconsistent category formatting",
    "records_affected": n_inconsistent_categories
})
```

```python
import pandas as pd

# Define the date range used by the retail sales dataset
start_date = "2024-01-01"
end_date = "2025-12-31"

# Generate one row for every calendar date
date_range = pd.date_range(
    start=start_date,
    end=end_date,
    freq="D"
)

# Build the date dimension
dim_date = pd.DataFrame({
    "date": date_range
})

# Add calendar attributes
dim_date["date_key"] = dim_date["date"].dt.strftime("%Y%m%d").astype(int)
dim_date["year"] = dim_date["date"].dt.year
dim_date["quarter"] = dim_date["date"].dt.quarter
dim_date["month"] = dim_date["date"].dt.month
dim_date["month_name"] = dim_date["date"].dt.month_name()
dim_date["week"] = dim_date["date"].dt.isocalendar().week.astype(int)
dim_date["day"] = dim_date["date"].dt.day
dim_date["day_of_week"] = dim_date["date"].dt.dayofweek + 1
dim_date["day_name"] = dim_date["date"].dt.day_name()

# Reorder columns
dim_date = dim_date[
    [
        "date_key",
        "date",
        "year",
        "quarter",
        "month",
        "month_name",
        "week",
        "day",
        "day_of_week",
        "day_name"
    ]
]

dim_date.head()
```
