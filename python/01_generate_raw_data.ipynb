```python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
```

```python
PROJECT_ROOT = Path.cwd().parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
CLEANED_DIR = PROJECT_ROOT / "data" / "cleaned"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

RAW_DIR.mkdir(parents=True, exist_ok=True)
CLEANED_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DIR)
```

```python
categories = {
    "Electronics": ["Mobile", "Laptop", "Audio", "Accessories"],
    "Home Appliances": ["Kitchen", "Cleaning", "Cooling"],
    "Fashion": ["Men", "Women", "Kids"],
    "Beauty": ["Skincare", "Haircare", "Fragrance"],
    "Sports": ["Fitness", "Outdoor", "Sportswear"],
    "Grocery": ["Beverages", "Snacks", "Pantry"],
    "Home & Living": ["Furniture", "Decor", "Bedding"]
}

brands = [
    "Samsung",
    "Apple",
    "Sony",
    "LG",
    "Philips",
    "Nike",
    "Adidas",
    "P&G",
    "L'Oreal",
    "Nestle",
    "IKEA",
    "Generic"
]
```

```python
cities = [
    "Riyadh",
    "Jeddah",
    "Dammam",
    "Khobar",
    "Mecca",
    "Medina",
    "Taif"
]

city_regions = {
    "Riyadh": "Central",
    "Jeddah": "Western",
    "Dammam": "Eastern",
    "Khobar": "Eastern",
    "Mecca": "Western",
    "Medina": "Western",
    "Taif": "Western"
}
```

```python
store_data = []

store_types = ["Mall", "Standalone", "Hypermarket"]

for i in range(1, 21):
    city = np.random.choice(cities)

    store_data.append({
        "store_id": f"S{i:03d}",
        "store_name": f"GulfMart {city} {i:02d}",
        "city": city,
        "region": city_regions[city],
        "store_type": np.random.choice(store_types)
    })

dim_store = pd.DataFrame(store_data)

dim_store.head()
```

```python
dim_store.to_csv(RAW_DIR / "dim_store.csv", index=False)
```
